3. Data Cleaning and derived  features

```python
df_clean = df.drop_duplicates(keep='first').copy()

# Only compute on_time where actual_delivery_date exists AND status is a completed one
df_clean['on_time'] = np.where(
    df_clean['actual_delivery_date'].notnull(),
    df_clean['actual_delivery_date'] <= df_clean['promised_delivery_date'],
    np.nan
)

df_clean['delay_days'] = (df_clean['actual_delivery_date'] - df_clean['promised_delivery_date']).dt.days
df_clean['transit_time_days'] = (df_clean['actual_delivery_date'] - df_clean['pickup_date']).dt.days
df_clean['pickup_lag_days'] = (df_clean['pickup_date'] - df_clean['booking_date']).dt.days
df_clean['cost_per_km'] = df_clean['freight_cost'] / df_clean['distance_km']

df_clean.to_csv('../data/shipments_clean.csv', index=False)
print(df_clean.shape)
```

**Data quality summary (Q4):**- 30 exact full-row duplicates dropped (0.6% of data)- booking_date / pickup_date: minor random gaps (~1.5-1.8%), excluded from lag calcs where null- Status/actual_delivery_date mismatch: 684 rows (13.6%) — Delivered/Delayed shipments missing actual_delivery_date — excluded from on-time% and delay_days rather than imputed, to avoid silently biasing headline metrics- Severe right-skew in freight_cost (skew ≈ 5.9) — used log-transform and median alongside mean throughout to avoid misleading averages- CARR_07 cost anomaly (see Q2 below) — likely a data/unit error, not genuine pricing

## Q2: Freight Cost vs Distance (relationship, and which carrier deviates?)

```python
import statsmodels.api as sm

X = sm.add_constant(df_clean['distance_km'])
y = df_clean['freight_cost']
model = sm.OLS(y, X, missing='drop').fit()
print(model.summary())
```

**Finding:** Distance alone is a weak predictor of freight cost (R² = 0.087).

```python
df_clean.groupby('mode')['freight_cost'].describe()
df_clean.groupby('mode')['cost_per_km'].describe()
```

```python
import statsmodels.formula.api as smf

model2 = smf.ols('freight_cost ~ distance_km + C(mode)', data=df_clean).fit()
print(model2.summary())
```

**Finding:** Adding transport mode improves the model (R² = 0.138) — cost is jointly driven by distance and mode.

```python
df_clean['log_cost'] = np.log(df_clean['freight_cost'])
model3 = smf.ols('log_cost ~ distance_km + C(mode)', data=df_clean).fit()
print(model3.summary())
```

**Finding:** A log-linear model (log(freight_cost) ~ distance + mode) fits far better, R² = 0.637 — consistent with the severe right-skew (skew ≈ 5.9) in raw freight_cost. This is the model used for carrier residual comparisons below.

```python
df_clean['predicted_cost'] = model2.predict(df_clean)
df_clean['cost_residual'] = df_clean['freight_cost'] - df_clean['predicted_cost']

carrier_residuals = df_clean.groupby('carrier_id')['cost_residual'].agg(['mean','count']).sort_values('mean')
print(carrier_residuals)
```

```python
carrier_residuals_full = df_clean.groupby('carrier_id')['cost_residual'].agg(['mean','median','std','count']).sort_values('mean')
print(carrier_residuals_full)
```

```python
carr07 = df_clean[df_clean['carrier_id']=='CARR_07'].sort_values('cost_residual', ascending=False)
print(carr07[['shipment_id','mode','distance_km','freight_cost','predicted_cost','cost_residual']].head(15))
```

```python
print((carr07['cost_residual'] > 10000).sum(), "out of", len(carr07))
```

```python
df_clean['predicted_log_cost'] = model3.predict(df_clean)
df_clean['log_cost_residual'] = df_clean['log_cost'] - df_clean['predicted_log_cost']

carrier_log_residuals = df_clean.groupby('carrier_id')['log_cost_residual'].agg(['mean','median','count']).sort_values('mean')
print(carrier_log_residuals)
```

```python
# Is this only in FTL, or across all modes for CARR_07?
carr07_by_mode = df_clean[df_clean['carrier_id']=='CARR_07'].groupby('mode')['freight_cost'].describe()
print(carr07_by_mode)

# Compare to other carriers' FTL costs for similar distance
ftl_others = df_clean[(df_clean['mode']=='FTL') & (df_clean['carrier_id']!='CARR_07')]
print(ftl_others['cost_per_km'].describe())

carr07_ftl = df_clean[(df_clean['mode']=='FTL') & (df_clean['carrier_id']=='CARR_07')]
print(carr07_ftl['cost_per_km'].describe())
```

**Q2 Answer:** All carriers except one cluster tightly (log-residual range: -18% to -11% vs the model's expected cost) — consistent, unremarkable pricing. **CARR_07 is a severe outlier**: costs average ~700–780% above expected, affecting 337 of 342 shipments (98.5%) — a systemic pattern, not a few outliers. Directly comparing cost-per-km confirms this: CARR_07's FTL cost-per-km averages ~249 vs ~25 for every other carrier — almost exactly a 10x factor, consistent across the full min/max range. A clean ~10x scaling strongly suggests a **systematic data/unit error** in CARR_07's cost records rather than genuine pricing — no carrier could commercially sustain 10x market rates. Recommend the data/ops team verify CARR_07's cost entries at the source before using this figure in any carrier renegotiation decision.

## Q3: Customer Delay Concentration (which customers, and why?)

```python
valid = df_clean.dropna(subset=['delay_days'])

customer_delay = valid.groupby('customer_id')['delay_days'].agg(['mean','median','count']).sort_values('mean', ascending=False)
print(customer_delay[customer_delay['count'] >= 5].head(15))
```

```python
top_customers = customer_delay[customer_delay['count'] >= 5].head(10).index.tolist()

top_cust_detail = valid[valid['customer_id'].isin(top_customers)].groupby(['customer_id','carrier_id']).size().unstack(fill_value=0)
print(top_cust_detail)

top_cust_region = valid[valid['customer_id'].isin(top_customers)].groupby(['customer_id','region']).size().unstack(fill_value=0)
print(top_cust_region)
```

```python
# Is delay simply proportional to volume? i.e., are "most delayed" customers just the ones who ship the most?
# valid.groupby('customer_id').size().corr(valid.groupby('customer_id')['delay_days'].mean())
cust_summary = valid.groupby('customer_id').agg(total_shipments=('delay_days','count'), avg_delay=('delay_days','mean'))
print(cust_summary['total_shipments'].corr(cust_summary['avg_delay']))
```

```python
top_cust_mode = valid[valid['customer_id'].isin(top_customers)].groupby(['customer_id','mode']).size().unstack(fill_value=0)
print(top_cust_mode)
```

**Q3 Answer:** The most delayed customers (e.g. CUST_079, CUST_119, CUST_026, averaging 1.6–1.8 days late) are **not** explained by a single dominant carrier — delay exposure is spread across 10+ carriers per customer, with only mild overrepresentation of CARR_07 for one customer (CUST_079: 28% of shipments vs ~7% baseline share). Delay is also not explained by shipment volume (correlation between total shipments and average delay = -0.16, negligible) or by transport mode mix (proportionally unremarkable). What stands out: these customers' shipments are heavily weighted toward Central, East, North, and West — the four weaker-performing regions from Q1 — with almost no shipments through South, the best-performing region. **Conclusion: delay concentration for these customers reflects their regional shipping mix, not a single carrier or customer-specific issue.** Recommend addressing the company-wide carrier reliability gap from Q1, which would benefit these customers as a side effect.

Tracking OTP by carrier weekly surfaces a carrier's decline in the same week it starts, well before a monthly aggregate would move enough to raise concern — and creates a direct audit trail for carrier contract reviews (tying into the Q2 cost-anomaly finding).**Secondary metric:** % of 'Delivered' shipments missing an actual_delivery_date — an early warning for the data pipeline issue found in Q4. A rising trend signals a tracking/system problem before it corrupts a month of delivery metrics.
